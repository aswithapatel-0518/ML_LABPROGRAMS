```python
# ============================================================
# EXPERIMENT 5
# XGBoost, LightGBM and SHAP - Placement Prediction
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip -q install xgboost lightgbm shap joblib


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import xgboost as xgb
import lightgbm as lgb
import shap
import joblib
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report
)


# ============================================================
# 3. UPLOAD PLACEMENT DATASET
# ============================================================

print("Upload your placement prediction CSV file:")

uploaded = files.upload()

# Automatically detect uploaded filename
filename = list(uploaded.keys())[0]

print("\nUploaded file:", filename)


# ============================================================
# 4. LOAD DATASET
# ============================================================

df = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. CHECK TARGET COLUMN
# ============================================================

print("\nDataset columns:")
print(df.columns.tolist())


# PlacementStatus is the target
target = "PlacementStatus"

if target not in df.columns:
    raise ValueError(
        f"Target column '{target}' was not found in the dataset."
    )


# ============================================================
# 6. PREPARE FEATURES
# ============================================================

X = df.drop(columns=[target]).copy()
y = df[target].copy()


# ------------------------------------------------------------
# Remove ID-type columns
# ------------------------------------------------------------

id_columns = []

for col in X.columns:

    if col.lower() in [
        "studentid",
        "student_id",
        "id"
    ]:

        id_columns.append(col)


if len(id_columns) > 0:

    X = X.drop(columns=id_columns)

    print("\nRemoved ID columns:", id_columns)


# ------------------------------------------------------------
# Remove Salary Package if present
# ------------------------------------------------------------
# Salary Package should not be used to predict placement
# because it is available only after placement.

salary_columns = []

for col in X.columns:

    if col.lower() in [
        "salarypackage",
        "salary_package",
        "salary",
        "package"
    ]:

        salary_columns.append(col)


if len(salary_columns) > 0:

    X = X.drop(columns=salary_columns)

    print("Removed leakage columns:", salary_columns)


# ============================================================
# 7. CONVERT CATEGORICAL VARIABLES
# ============================================================

categorical_columns = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()


print("\nCategorical columns:")

print(categorical_columns)


# Convert categorical columns using one-hot encoding
if len(categorical_columns) > 0:

    X = pd.get_dummies(
        X,
        columns=categorical_columns,
        drop_first=True
    )


# ============================================================
# 8. CONVERT ALL FEATURES TO NUMERIC
# ============================================================

X = X.apply(
    pd.to_numeric,
    errors="coerce"
)


# Replace missing values
X = X.fillna(X.median())


# Make sure all columns are numeric
X = X.astype(float)


print("\nFinal feature shape:", X.shape)


# ============================================================
# 9. CONVERT TARGET TO 0 AND 1
# ============================================================

print("\nTarget values before conversion:")
print(y.value_counts())


if y.dtype == "object":

    # Automatically convert common Yes/No placement labels

    unique_values = y.dropna().unique()

    if set(unique_values).issubset(
        {"Yes", "No"}
    ):

        y = y.map({
            "No": 0,
            "Yes": 1
        })

    elif set(unique_values).issubset(
        {"Placed", "Not Placed"}
    ):

        y = y.map({
            "Not Placed": 0,
            "Placed": 1
        })

    else:

        # Generic binary conversion
        values = list(unique_values)

        if len(values) == 2:

            y = y.map({
                values[0]: 0,
                values[1]: 1
            })

        else:

            raise ValueError(
                "PlacementStatus must contain two classes."
            )


y = y.astype(int)


print("\nTarget values after conversion:")
print(y.value_counts())


# ============================================================
# 10. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# 70% Training
# 15% Validation
# 15% Testing

Xtr, Xtmp, ytr, ytmp = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)


Xva, Xte, yva, yte = train_test_split(
    Xtmp,
    ytmp,
    test_size=0.50,
    random_state=42,
    stratify=ytmp
)


print("\nDataset split:")
print("Training   :", Xtr.shape[0])
print("Validation :", Xva.shape[0])
print("Testing    :", Xte.shape[0])


# ============================================================
# 11. XGBOOST MODEL
# ============================================================

print("\nTraining XGBoost...")

xg = xgb.XGBClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="auc",
    early_stopping_rounds=50,
    random_state=42,
    n_jobs=-1
)


xg.fit(
    Xtr,
    ytr,
    eval_set=[(Xva, yva)],
    verbose=False
)


print("XGBoost training completed!")


# ============================================================
# 12. LIGHTGBM MODEL
# ============================================================

print("\nTraining LightGBM...")

lg = lgb.LGBMClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    num_leaves=15,
    random_state=42,
    verbose=-1,
    n_jobs=-1
)


lg.fit(
    Xtr,
    ytr,
    eval_set=[(Xva, yva)],
    callbacks=[
        lgb.early_stopping(
            50,
            verbose=False
        )
    ]
)


print("LightGBM training completed!")


# ============================================================
# 13. MODEL EVALUATION
# ============================================================

print("\n================================================")
print("MODEL PERFORMANCE")
print("================================================")


for name, model in [
    ("XGBoost", xg),
    ("LightGBM", lg)
]:

    # Probability of Placement = 1
    p = model.predict_proba(Xte)[:, 1]

    auc = roc_auc_score(
        yte,
        p
    )

    pr_auc = average_precision_score(
        yte,
        p
    )

    print(
        f"{name:<10} "
        f"AUC={auc:.4f}   "
        f"PR-AUC={pr_auc:.4f}"
    )


# ============================================================
# 14. CLASSIFICATION REPORT - XGBOOST
# ============================================================

print("\n================================================")
print("XGBOOST CLASSIFICATION REPORT")
print("================================================")

xg_predictions = xg.predict(Xte)

print(
    classification_report(
        yte,
        xg_predictions
    )
)


# ============================================================
# 15. CLASSIFICATION REPORT - LIGHTGBM
# ============================================================

print("\n================================================")
print("LIGHTGBM CLASSIFICATION REPORT")
print("================================================")

lg_predictions = lg.predict(Xte)

print(
    classification_report(
        yte,
        lg_predictions
    )
)


# ============================================================
# 16. SHAP ANALYSIS
# ============================================================

print("\nGenerating SHAP explanations...")


# Create SHAP Tree Explainer
explainer = shap.TreeExplainer(xg)


# Calculate SHAP values
sv = explainer(Xte)


# ============================================================
# 17. SHAP SUMMARY PLOT
# ============================================================

print("\nSHAP Summary Plot:")

shap.summary_plot(
    sv,
    Xte
)


# ============================================================
# 18. SHAP WATERFALL PLOT
# ============================================================

print("\nSHAP Waterfall Plot for First Test Student:")

shap.plots.waterfall(
    sv[0]
)


# ============================================================
# 19. SAVE XGBOOST MODEL
# ============================================================

joblib.dump(
    xg,
    "placement_model.joblib"
)


print("\n================================================")
print("XGBoost model saved as:")
print("placement_model.joblib")
print("================================================")


# ============================================================
# 20. DOWNLOAD MODEL
# ============================================================

files.download(
    "placement_model.joblib"
)


print("\nExperiment 8 completed successfully!")
```
